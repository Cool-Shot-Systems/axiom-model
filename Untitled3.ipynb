```python
# --- COOL SHOT SYSTEMS: AXIOM LAUNCH ---
# Diamond Stack | T4 Optimized | Llama 3.1

import os
import sys
import torch
from unsloth import FastLanguageModel
from trl import SFTTrainer
from transformers import TrainingArguments
from datasets import Dataset

# 1. VERIFY STACK
print("🔍 SYSTEM CHECK...")
try:
    import unsloth
    import unsloth_zoo
    print(f"   - Unsloth: {unsloth.__version__}")
    print("✅ SYSTEM GREEN.")
except ImportError as e:
    print(f"❌ ERROR: {e}")
    print("   Did you run Step 1 and then Restart?")
    sys.exit()

# 2. FILE UPLOAD (Since restart wiped them)
print("\n📂 UPLOAD YOUR 5 FILES NOW...")
required_files = ["coding_philosophy.txt", "biblical_principles.txt", "founder_bio.txt", "economics_notes.txt", "cool_shot_manifesto.txt"]
dataset_text = []

try:
    from google.colab import files
    uploaded = files.upload()
    for fname in uploaded.keys():
        text = uploaded[fname].decode("utf-8")
        formatted_text = f"### Title: {fname}\n### Content:\n{text}\n<|end_of_text|>"
        dataset_text.append({"text": formatted_text})
        print(f"   - Ingested: {fname}")
except:
    print("⚠️ Kaggle Mode: Checking local directory...")
    # Fallback for Kaggle if files were already uploaded to input
    import glob
    for fname in glob.glob("*.txt"):
        with open(fname, "r", encoding="utf-8") as f:
            text = f.read()
            formatted_text = f"### Title: {fname}\n### Content:\n{text}\n<|end_of_text|>"
            dataset_text.append({"text": formatted_text})
            print(f"   - Ingested: {fname}")

if len(dataset_text) == 0:
    print("❌ STOP: No files found.")
    sys.exit()

# 3. LOAD BRAIN (T4 Safe Mode)
print("\n🧠 LOADING AXIOM...")
# We use the new model name, but with the old Unsloth logic
model_name = "unsloth/Meta-Llama-3.1-8B-bnb-4bit"

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = model_name,
    max_seq_length = 1024, # CRITICAL: 1024 fits on T4. 2048 crashes.
    dtype = torch.float16, # CRITICAL: T4 needs Float16.
    load_in_4bit = True,
)

model = FastLanguageModel.get_peft_model(
    model, r=16,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"],
    lora_alpha=16, lora_dropout=0, bias="none",
    use_gradient_checkpointing="unsloth", random_state=3407,
)

# 4. TRAIN
print(f"\n✨ CLONING DATA ({len(dataset_text)} files -> {len(dataset_text)*20})...")
dataset_text = dataset_text * 20
dataset = Dataset.from_list(dataset_text)

print("🚀 LAUNCHING TRAINING...")
trainer = SFTTrainer(
    model = model,
    tokenizer = tokenizer,
    train_dataset = dataset,
    dataset_text_field = "text",
    max_seq_length = 1024,
    dataset_num_proc = 2,
    packing = False,
    args = TrainingArguments(
        per_device_train_batch_size = 1, # Safe size
        gradient_accumulation_steps = 8, # Compensate for small batch
        warmup_steps = 5,
        max_steps = 60,
        learning_rate = 2e-4,
        fp16 = True,                     # T4 Supported
        bf16 = False,                    # T4 Unsupported
        logging_steps = 1,
        output_dir = "outputs",
        optim = "adamw_8bit",
        weight_decay = 0.01,
        lr_scheduler_type = "linear",
        seed = 3407,
    ),
)
trainer.train()

# 5. SAVE & DOWNLOAD
print("\n💾 SAVING AXIOM...")
model.save_pretrained("axiom_model")
tokenizer.save_pretrained("axiom_model")

print("📦 ZIPPING...")
os.system("zip -r axiom_model.zip axiom_model")

print("\n⬇️ DOWNLOADING...")
try:
    from google.colab import files
    files.download('axiom_model.zip')
except:
    print("✅ Kaggle: File saved as 'axiom_model.zip'.")
```
